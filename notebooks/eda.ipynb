```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
customer=pd.read_csv("customer.csv")
```

```python
customer.sample(5)
```

```python
customer.shape
```

```python
customer.info()
```

Change data type of Total charges to float

```python
customer['TotalCharges']=pd.to_numeric(customer['TotalCharges'],errors='coerce')
```

```python
customer.isnull().sum()
```

```python
customer['TotalCharges']=customer['TotalCharges'].fillna(0,inplace=True)
```

```python
customer.info()
```

```python
customer.isnull().sum()
```

```python
customer.describe()
```

```python
customer.describe(include='object')
```

```python
customer=customer.drop("customerID",axis=1)
```

Customer ID wont be of help so we remove that.

So we can see that the dataset contains no missing values. 

```python
customer.duplicated().sum()
```

```python
customer[customer.duplicated()]
```

```python
customer=customer.drop_duplicates()
```

```python
customer.corr(numeric_only='true')
```

```python
#change the data type of SeniorCitizen to category for proper analysis
customer['SeniorCitizen']=customer['SeniorCitizen'].astype('category')
```

Univariate Analysis

```python
sns.countplot(x='Churn',data=customer)
plt.show()
```

```python
for col in customer:
    print(customer[col].value_counts())
    print("-------------------")
```

```python
for col in customer.select_dtypes('number'):
    print(col,"--",customer[col].skew())
```

```python
for col in customer.select_dtypes('number'):
    sns.kdeplot(data=customer,x=col,hue='Churn')
    plt.show()
```

The plot shows that as the tenure increases, people are less likely to churn .
As charges increases both churn and not churn decreases continuosly but number of people who dont churn is high.

```python
c1=['gender','SeniorCitizen','Partner','Dependents']
for col in c1:
    customer.groupby([col, 'Churn']).size().unstack().plot(kind='bar')
    plt.title(f"{col} vs Churn")
    plt.show()
```

```python
for col in customer.select_dtypes('number'):
    sns.boxplot(x=customer[col])
    plt.show()
```

There are no outliers .

Bivariate Analysis

```python
plt.figure(figsize=(10, 6))
sns.barplot(x='PaymentMethod',y='TotalCharges',hue='Churn',data=customer)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(x='InternetService',y='tenure',hue='Churn',data=customer)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(x='PhoneService',y='tenure',hue='Churn',data=customer)
plt.show()
```
